```python
orders["delivery_time_days"] = (
    orders["order_delivered_customer_date"] -
    orders["order_purchase_timestamp"]
).dt.days
#if we dont write dt.days, exact delivery is calculated (with timestamps too, instead of just days), like this "8 days 10:28:40"
```

```python
orders["estimated_delivery_days"] = (
    orders["order_estimated_delivery_date"] -
    orders["order_purchase_timestamp"]
).dt.days.astype("Int64")
```

```python
orders["estimated_delivery_days"]
```

```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"] -
    orders["order_estimated_delivery_date"]
).dt.days.astype("Int64")
```

```python
orders["delivery_delay_days"]
```

```python
def time_of_day(hour):
    if 0 <= hour < 7:
        return "Dawn"
    elif 7 <= hour < 13:
        return "Morning"
    elif 13 <= hour < 19:
        return "Afternoon"
    else:
        return "Night"

orders["time_of_day"] = orders["order_purchase_timestamp"].dt.hour.apply(time_of_day)
```

```python
products.duplicated().sum()
```

```python
products.rename(columns={"product category": "product_category"}, inplace=True)
```

```python
products["product_category"] = products["product_category"].fillna("Unknown")
```

```python
num_cols = [
    "product_name_length",
    "product_description_length",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in num_cols:
    products[col] = products[col].fillna(products[col].median())
```

```python
products[num_cols] = products[num_cols].astype(int)
```

```python
(products["product_weight_g"] == 0).sum()
```

```python
products[products["product_weight_g"] == 0]
```

```python
#all are of same category, so we cant rlly use groupby unless there are more than 4 items in same category, so we check them

products["product_category"].value_counts()
```

```python
import numpy as np
```

```python
# Replace 0 with nan
products["product_weight_g"] = products["product_weight_g"].replace(0, np.nan)

# Fill with the median weight of each category
products["product_weight_g"] = products["product_weight_g"].fillna(
    products.groupby("product_category")["product_weight_g"].transform("median")
)

# Convert to integer
products["product_weight_g"] = products["product_weight_g"].astype(int)
```

```python
(products["product_weight_g"] == 0).sum()
```

```python
products["product_volume_cm3"] = (
    products["product_length_cm"] *
    products["product_width_cm"] *
    products["product_height_cm"]
)
```
